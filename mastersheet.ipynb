```python
current_customer = None
current_category = None

pending_label = None

for index, row in df.iterrows():

    first_col = str(row.iloc[0]).strip()
    bill_date = row.iloc[1]
    product_name = str(row.iloc[2]).strip()

    # Blank row
    if first_col == "nan" and pd.isna(bill_date):
        continue

    # Total row
    if product_name.lower() == "total":
        continue

    # Transaction row
    if pd.notna(bill_date):

        # Header ko ignore karo
        if first_col.lower() == "bill no":
            continue

        print(
            index,
            "| Customer:", current_customer,
            "| Category:", current_category,
            "| Bill:", first_col
        )

    # Label row
    else:

        if pending_label is None:
            pending_label = first_col

        else:
            # Two consecutive labels:
            # first = Customer
            # second = Category

            current_customer = pending_label
            current_category = first_col

            pending_label = None

            print(
                index,
                "| NEW CUSTOMER:", current_customer,
                "| NEW CATEGORY:", current_category
            )
```

```python
current_customer = None
current_category = None

skip_next_label = False

for index in range(len(df)):

    row = df.iloc[index]

    first_col = row.iloc[0]
    bill_date = row.iloc[1]
    product_name = row.iloc[2]

    # 1. Blank row
    if pd.isna(first_col) and pd.isna(bill_date) and pd.isna(product_name):
        continue

    # 2. Total row
    if str(product_name).strip().lower() == "total":
        continue

    # 3. Header row
    if str(first_col).strip().lower() == "bill no":
        continue

    # 4. Transaction row
    if pd.notna(bill_date):

        print(
            index,
            "| Customer:", current_customer,
            "| Category:", current_category,
            "| Bill:", str(first_col).strip()
        )

    # 5. Label row
    else:

        # Agar ye label already customer ke saath category ke
        # roop mein use ho chuka hai, to ise skip karo
        if skip_next_label:
            skip_next_label = False
            continue

        label = str(first_col).strip()

        # Next meaningful row find karo
        next_index = index + 1

        while next_index < len(df):

            next_row = df.iloc[next_index]

            next_first = next_row.iloc[0]
            next_date = next_row.iloc[1]
            next_product = next_row.iloc[2]

            # Blank row ignore
            if (
                pd.isna(next_first)
                and pd.isna(next_date)
                and pd.isna(next_product)
            ):
                next_index += 1
                continue

            # Total row ignore
            if str(next_product).strip().lower() == "total":
                next_index += 1
                continue

            break

        # Next row bhi label hai
        if next_index < len(df):

            next_first = df.iloc[next_index].iloc[0]
            next_date = df.iloc[next_index].iloc[1]

            if pd.notna(next_first) and pd.isna(next_date):

                # First label = Customer
                current_customer = label

                # Second label = Category
                current_category = str(next_first).strip()

                # Next row ko dobara category mat banana
                skip_next_label = True

                print(
                    index,
                    "| NEW CUSTOMER:",
                    current_customer,
                    "| NEW CATEGORY:",
                    current_category
                )

            # Next row transaction hai
            elif pd.notna(next_date):

                current_category = label

                print(
                    index,
                    "| NEW CATEGORY:",
                    current_category
                )
```

```python
def clean_sheet(df):

    current_customer = None
    current_category = None

    skip_next_label = False

    cleaned_rows = []

    for index in range(len(df)):

        row = df.iloc[index]

        first_col = row.iloc[0]
        bill_date = row.iloc[1]
        product_name = row.iloc[2]
        ri = row.iloc[3]
        qty = row.iloc[4]
        rate = row.iloc[5]

        # -------------------------
        # 1. Blank row
        # -------------------------
        if (
            pd.isna(first_col)
            and pd.isna(bill_date)
            and pd.isna(product_name)
        ):
            continue

        # -------------------------
        # 2. Total row
        # -------------------------
        if str(product_name).strip().lower() == "total":
            continue

        # -------------------------
        # 3. Header row
        # -------------------------
        if str(first_col).strip().lower() == "bill no":
            continue

        # -------------------------
        # 4. Transaction row
        # -------------------------
        if pd.notna(bill_date):

            cleaned_rows.append([
                current_customer,
                current_category,
                first_col,
                bill_date,
                product_name,
                ri,
                qty,
                rate
            ])

        # -------------------------
        # 5. Label row
        # -------------------------
        else:

            # Already used as category
            if skip_next_label:
                skip_next_label = False
                continue

            label = str(first_col).strip()

            # Find next meaningful row
            next_index = index + 1

            while next_index < len(df):

                next_row = df.iloc[next_index]

                next_first = next_row.iloc[0]
                next_date = next_row.iloc[1]
                next_product = next_row.iloc[2]

                # Blank row
                if (
                    pd.isna(next_first)
                    and pd.isna(next_date)
                    and pd.isna(next_product)
                ):
                    next_index += 1
                    continue

                # Total row
                if str(next_product).strip().lower() == "total":
                    next_index += 1
                    continue

                break

            # -------------------------
            # New Customer + Category
            # -------------------------
            if next_index < len(df):

                next_first = df.iloc[next_index].iloc[0]
                next_date = df.iloc[next_index].iloc[1]

                # Next row is another label
                if pd.notna(next_first) and pd.isna(next_date):

                    current_customer = label
                    current_category = str(next_first).strip()

                    skip_next_label = True

                # -------------------------
                # New Category
                # -------------------------
                elif pd.notna(next_date):

                    current_category = label

    # -------------------------
    # Create DataFrame
    # -------------------------

    columns = [
        "Customer Name",
        "Product Category",
        "Bill No",
        "Bill Date",
        "Product Name",
        "R/I",
        "Qty",
        "Rate"
    ]

    cleaned_df = pd.DataFrame(
        cleaned_rows,
        columns=columns
    )

    return cleaned_df
```

```python
all_cleaned = []

for sheet_name, raw_df in excel_data.items():

    cleaned_df = clean_sheet(raw_df)

    # Sheet ka naam bhi save kar lete hain
    cleaned_df["Source Sheet"] = sheet_name

    all_cleaned.append(cleaned_df)

    print(
        sheet_name,
        "→",
        len(cleaned_df),
        "clean rows"
    )
```

```python
all_cleaned = []

for sheet_name, raw_df in excel_data.items():

    cleaned_df = clean_sheet(raw_df)

    # Exact duplicate rows remove
    cleaned_df = cleaned_df.drop_duplicates().reset_index(drop=True)

    # Source sheet preserve karna useful rahega
    cleaned_df["Source Sheet"] = sheet_name

    all_cleaned.append(cleaned_df)

    print(
        sheet_name,
        "→",
        len(cleaned_df),
        "clean unique rows"
    )
```

```python
final_df = pd.concat(
    all_cleaned,
    ignore_index=True
)

print("Final rows:", len(final_df))
print("Final columns:", final_df.columns.tolist())
```
